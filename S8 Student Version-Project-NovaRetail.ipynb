### Punto-biserial

```python


# Calcular correlación entre variables relevantes
from scipy.stats import pointbiserialr
binarias=['miembro_premium','abandono']
def corr_point_biserial(df, binarias, ingreso_anual):
    for col in binarias:
        coef, p_value = pointbiserialr(df[col],df['ingreso_anual'])
        print (f'\nCorrelacion entre: {col} y {ingreso_anual}')
        print(f'Point-Biserial:{coef:.4f}')
corr_point_biserial(df, binarias,'ingreso_anual')


```

**Observaciones Punto-biserial** 

`miembro_premium` vs `ingreso_anual`
- Dirección: Positiva (0.0931). Los clientes con membresía premium tienden a presentar un ingreso anual ligeramente mayor.
- Magnitud: Muy baja. La asociación es débil, por lo que la membresía premium, por sí sola, no parece estar fuertemente relacionada con el ingreso anual.

`abandono` vs `ingreso_anual`
- Dirección: Negativa (-0.0028). Los clientes que abandonan la plataforma tienden a presentar un ingreso anual ligeramente menor.
- Magnitud: Prácticamente nula. El coeficiente es muy cercano a cero, indicando que no existe una relación lineal relevante entre el abandono y el ingreso anual.


- **En resumen:** La correlación punto-biserial muestra que ni la membresía premium ni el abandono presentan una asociación fuerte con el ingreso anual. Aunque la membresía premium tiene una relación positiva, esta es muy débil, mientras que el abandono presenta una relación negativa prácticamente inexistente. En comparación con variables como compras_mes, estas variables binarias tienen un impacto mucho menor sobre el ingreso anual.

### V de Cramér

```python
# Función para calcular V de Cramér
from scipy.stats import chi2_contingency
def cramer_v (df,tipo_dispositivo,region):
    tabla = pd.crosstab(df['tipo_dispositivo'],df['region'])
    chi2, p_value, dof, expected = chi2_contingency(tabla)
    n= tabla.values.sum()
    v= np.sqrt(chi2/n*(min(tabla.shape)-1))
    return v
```

```python
# Aplicar V de Cramér en variables relevantes


coef_cramer = cramer_v (df,'tipo_dispositivo','region')
print ('\nCorrelación entre: region y tipo dispositivo')
print(f'V de cramer: {coef_cramer:.3f}')

```

El coeficiente V de Cramér es 0.025, lo que indica una asociación muy débil o prácticamente nula entre las variables `tipo_dispositivo` y `region`.
Esto sugiere que el tipo de dispositivo utilizado por los clientes no depende de la región en la que se encuentran, ya que no existe una relación relevante entre ambas variables.
En consecuencia, estas variables parecen aportar información independiente y no presentan una asociación categórica significativa

- **En Resumen:** El análisis mediante V de Cramér muestra que tipo_dispositivo y region son variables prácticamente independientes, por lo que la distribución del dispositivo utilizado es similar entre las diferentes regiones. Este resultado indica que no existe una asociación categórica importante entre ambas variables.
